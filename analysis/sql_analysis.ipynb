```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
engine = create_engine("mssql+pyodbc://Chand\\SQLEXPRESS/AnalyticsDB?driver=ODBC+Driver+18+for+SQL+Server&trusted_connection=yes&TrustServerCertificate=yes")
```

```python
query = """
SELECT 
p.Name,
sum(s.SalesAmount) as Revenue
FROM
sales s
LEFT JOIN products as p ON s.ProductID = p.ProductID
GROUP BY p.Name
ORDER BY Revenue desc
"""
df_product_revenue = pd.read_sql(query, engine)
df_product_revenue.head()
```

```python
query = """
SELECT 
c.Country,
sum(s.SalesAmount) as Revenue
FROM
sales as s
LEFT JOIN customers as c ON s.CustomerID = c.CustomerID
GROUP BY c.Country
ORDER BY Revenue desc;
"""

df_country_revenue = pd.read_sql(query, engine)
df_country_revenue.head()
```

```python
query = """ 
SELECT
c.CustomerID,
CONCAT(c.FirstName,' ', c.LastName) as CustomerName,
sum(s.SalesAmount) as Revenue
FROM
sales as s
LEFT JOIN customers as c ON s.CustomerID = c.CustomerID
GROUP BY c.CustomerID, c.FirstName, c.LastName
ORDER BY Revenue desc
"""

df_top_customers = pd.read_sql(query, engine)
df_top_customers.head()
```

```python
query = """ 
SELECT 
c.Country,
p.Name,
sum(s.SalesAmount) as Revenue
FROM
sales as s
LEFT JOIN products as p ON s.ProductID = p.ProductID
LEFT JOIN customers as c ON s.CustomerID = c.CustomerID
GROUP BY c.Country, p.Name
ORDER BY c.Country asc, Revenue desc
"""

df_product_country = pd.read_sql(query, engine)
df_product_country.head()
```

```python
query = """
SELECT
DISTINCT p.Category,
sum(s.SalesAmount) as Revenue
FROM sales as s
LEFT JOIN products as p ON s.ProductID = p.ProductID
GROUP BY p.Category
ORDER BY Revenue desc
"""

df_category_sales = pd.read_sql(query, engine)
df_category_sales.head()
```

```python
query = """ 
SELECT
FORMAT(OrderDate, 'yyyy-MM') as YearMonth,
sum(SalesAmount) as Revenue
FROM sales
GROUP BY FORMAT(OrderDate, 'yyyy-MM')
ORDER BY YearMonth"""

df_monthly_sales = pd.read_sql(query, engine)
df_monthly_sales.head()
```

```python
query = """ 
SELECT
DISTINCT p.Name,
sum(s.Quantity) as TotalSalesQnty,
sum(s.SalesAmount) as Revenue
FROM
sales as s
LEFT JOIN products as p ON s.ProductID = p.ProductID
GROUP BY p.Name
ORDER BY Revenue desc
"""

df_product_sales = pd.read_sql(query, engine)
df_product_sales.head()
```
